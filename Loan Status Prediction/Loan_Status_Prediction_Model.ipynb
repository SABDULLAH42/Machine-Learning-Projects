```python
X = df[["Married", "ApplicantIncome", "Education", "LoanAmount", "Credit_History"]]
X

# X[0,3500,1,100,1] that input will be used for model prediction
```

```python
y = df[["Loan_Status"]]
```

```python
y
```

### Preprocessing

At first we do Preprocessing using Laber Encoder for change this 

text features to numeric ones so that ther should be

1 and 0 for X, 1 and 0 for y

```python
from sklearn import preprocessing
# labelencoder = preprocessing.LabelEncoder()
```

```python
from sklearn.preprocessing import LabelEncoder
labelencoder = LabelEncoder()
```

```python
y["Loan_Status"] = labelencoder.fit_transform(y["Loan_Status"])
```

```python
y
```

```python
labelencoder.classes_

# N >> 0 and Y >>1
```

```python
X["Credit_History"][0]
```

```python
X["Married"] = labelencoder.fit_transform(X["Married"])
```

```python
X
```

```python
X["Education"] = labelencoder.fit_transform(X["Education"])
X
```

### Scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X = scaler.fit_transform(X)
X
```

```python
import joblib
joblib.dump(scaler, "Scaler.pkl")
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3)
```

```python
from sklearn.metrics import accuracy_score
```

```python
def modelperformance(prediction):
    print("Accuracy score in model is {}".format(accuracy_score(y_test, prediction)))
```

```python
from sklearn.model_selection import GridSearchCV

# It will help in hyperparametr tuning side
```

```python
from sklearn.linear_model import LogisticRegression
log_model = LogisticRegression()
log_model.fit(X_train, y_train.values.ravel())
```

```python
log_predictions = log_model.predict(X_test)
```

```python
modelperformance(log_predictions)
```

```python
from sklearn.neighbors import KNeighborsClassifier
param_grid = {"n_neighbors": [3, 5, 7, 8, 6],
              "weights": ["uniform", "distance"]}

gridkn = GridSearchCV(KNeighborsClassifier(), param_grid)
gridkn.fit(X_train, y_train.values.ravel())
```

```python
gridkn.best_params_
```

```python
knpreds = gridkn.predict(X_test)
```

```python
modelperformance(knpreds)
```

```python
from sklearn.svm import SVC
svm = SVC()
param_grid_svc = {"C": [0.04, 0.01, 0.02, 0.1, 0.5], "kernel":["linear", "rbf", "poly"]}
gridsvc = GridSearchCV(svm, param_grid_svc)
```

```python
gridsvc.fit(X_train, y_train.values.ravel())
```

```python
gridsvc.best_params_
```

```python
pred_svc = gridsvc.predict(X_test)
modelperformance(pred_svc)
```

```python
joblib.dump(gridsvc, "model.pkl")
```
